Проблемы продукта:
* операционные
* стратегические

Операционные проблемы:
* 8,11% доставленных заказов были пришли с задержкой
* в половине заказов клиент платит более 18% от суммы на доставку
* 0,63% заказов были отменены из-за бага
* 0,61% заказов стали недоступными

Стратегическая: у продукта retention следующего после первой покупки месяца равен околонулевым значениям по всем когортам. Даже если отфильтровать клиентов с хорошим первым опытом: заказ доставлен в срок и доля доставки менее 10% — retention остается околонулевым. Это говорит об отсутствии product/market fit: клиенты совершают первую покупку, но не возвращаются

Продуктовые метрики для мониторинга:
* месячный объем продаж по доставленным заказам
* количество уникальных клиентов по месяцам оформления заказа (только доставленные заказы)
* конверсия новых пользователей в первый заказ
* retention на следующий месяц после первой покупки
* объем продаж, который генерирует один клиент в месяц (только доставленные заказы)

Учитывая, что объем продаж и количество клиентов растет, retention околонулевой, а объем продаж на клиента почти всегда 140-180 R$, можно сказать, что маркетплейс вкладывается в агрессивное привлечение новой аудитории, при этом сохраняя околонулевую возвращаемость после первой покупки

Из трех предложенных гипотез приоритетной по фреймворку ICE является первая: "Если исправим баг в системе процессинга заказов, то клиентам не придётся сталкиваться с проблемой отмены заказа, вследствие чего количество доставленных заказов увеличится. Считаем, что мы таким образом избавимся от всех отмен"

Метрики A/B теста:
* целевые: доля доставленных заказов; доля отменённых заказов
* прокси: доля заказов с отсутствующим временем подтверждения
* guardrail: доля заказов, доставленных позже ожидаемой даты

При ожидаемом эффекте, близком к полному устранению отмен, вероятности равной 0.05 ошибочно найти значимый эффект и вероятности равной 0.2 упустить значимый эффект нужно по 1 241 заказу на группу. При медианном потоке 148 заказов в день эксперимент займет около 17 дней плюс время пока все эти заказы дойдут

Рекомендации по продукту:
* провести A/B тест первой гипотезы
* провести PMF-исследование
* постепенно перейти от стратегии привлечения к стратегии удержания
